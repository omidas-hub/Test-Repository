```python
import pandas as pd
from google.colab import files
```

```python
file = "/content/drive/MyDrive/AFS Data /Tree Plantation.xlsx"
```

```python
general = pd.read_excel(file, sheet_name="main_info")
```

```python
species = pd.read_excel(file, sheet_name="species")
```

```python
ben = pd.read_excel(file, sheet_name="beneficiary")
```

```python
labor = pd.read_excel(file, sheet_name="labor")
```

```python
mask = general["Type of Plantation"].str.strip().str.lower() == "other"
```

```python
general.loc[mask, "Type of Plantation"] = (
    general.loc[mask, "If other, please specify"]
    .str.strip()
    .str.title()
)
```

```python
general["Type of Plantation"] = general["Type of Plantation"].str.strip().str.title()
```

```python
general.drop(columns=["If other, please specify"], inplace=True)
```

```python
mask = species["Name of Species"].str.strip().str.lower() == "other"
```

```python
species.loc[mask, "Name of Species"] = (
    species.loc[mask, "If other, please specify"]
    .str.strip()
    .str.title()
)
```

```python
species["Name of Species"] = species["Name of Species"].str.strip().str.title()
```

```python
species.drop(columns=["If other, please specify"], inplace=True)
```

```python
general_sp = general.merge(
  species,
  left_on="_index",
  right_on="_parent_index",
  how="left"
)
```

```python
final = general_sp.merge(
    ben,
    left_on="_index",
    right_on="_parent_index",
    how="left"
)
```

```python
parent_fields = [
    "Number of Sites",
    "Total  Area",
    "Total Number of Households",
    "Number of Female-headed Household/s",
    "Male",
    "Female"
]
```

```python
for col in parent_fields:
  final.loc[final.duplicated(subset="_index"), col] = pd.NA
```

```python
existing_parent_fields = [
    col
    for col in parent_fields
    if col in final.columns
]
```

```python
output_path = "/content/drive/MyDrive/AFS Dashboard/Tree_Merged.xlsx"
```

```python
final.to_excel(output_path, index=False)
```

```python
Tree_labor = general.merge(
    labor,
    left_on="_index",
    right_on="_parent_index",
    how="left"
)
```

```python
output_path = "/content/drive/MyDrive/AFS Dashboard/Tree_labor.xlsx"
```

```python
Tree_labor.to_excel(output_path, index=False)
```
